```python
import numpy as np

# Données
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

# Initialisation
a, b = 0.0, 0.0
lr = 0.1
epochs = 2000
epsilon = 1e-6
previous_cost = float('inf')

# Entraînement
for i in range(epochs):
    y_pred = a * X + b
    
    current_cost = (1/len(X)) * sum((y - y_pred)**2)
    
    if abs(previous_cost - current_cost) < epsilon:
        break
        
    previous_cost = current_cost
    
    da = (-2/len(X)) * sum(X * (y - y_pred))
    db = (-2/len(X)) * sum(y - y_pred)
    
    a = a - lr * da
    b = b - lr * db

print(f"y = {a:.2f}x + {b:.2f}")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# 1. Préparation des données
# X doit toujours être une matrice 2D (d'où le .reshape(-1, 1))
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 5])

# 2. Initialisation du modèle
model = LinearRegression()

# 3. Entraînement (Le "fit")
# C'est ici que l'algorithme calcule a et b automatiquement
model.fit(X, y)

# 4. Prédiction
# On demande au modèle de prédire pour une nouvelle valeur (ex: 6)
nouvelle_valeur = np.array([[6]])
prediction = model.predict(nouvelle_valeur)

print(f"Pente (a) : {model.coef_[0]:.2f}")
print(f"Ordonnée à l'origine (b) : {model.intercept_:.2f}")
print(f"Prédiction pour x=6 : {prediction[0]:.2f}")

# (Optionnel) Visualisation
plt.scatter(X, y, color='blue') # Points réels
plt.plot(X, model.predict(X), color='red') # Ligne de régression
plt.title("Régression Linéaire avec Scikit-Learn")
plt.show()
```

```python
import numpy as np

# Données : [Surface, Nb_Chambres]
X = np.array([[50, 2], [80, 3], [100, 4], [30, 1], [120, 5]])
y = np.array([200, 310, 390, 130, 480])

# 1. Initialisation
n_samples, n_features = X.shape
W = np.zeros(n_features) # Un poids pour chaque feature [0, 0]
b = 0
lr = 0.0001
epochs = 1000

for i in range(epochs):
    # 2. Prédiction (Produit scalaire entre X et W)
    y_pred = np.dot(X, W) + b
    
    # 3. Fonction de Coût (MSE)
    cost = (1/n_samples) * np.sum((y_pred - y)**2)
    
    # 4. Gradients (Calculés sous forme matricielle)
    dw = (2/n_samples) * np.dot(X.T, (y_pred - y))
    db = (2/n_samples) * np.sum(y_pred - y)
    
    # 5. Mise à jour
    W = W - lr * dw
    b = b - lr * db

print(f"Poids (W): {W}, Biais (b): {b:.2f}")
```

```python
import numpy as np

# Données (Non-linéaires : courbe)
X = np.array([1, 2, 3, 4, 5])
y = np.array([1.2, 4.5, 8.8, 16.2, 24.8]) # Courbe style x^2

# 1. STEP POLYNOMIAL : Transformer X en [x, x^2]
X_poly = np.column_stack((X, X**2))

# 2. STEP LINEAR MULTIPLE (On réutilise la même logique que plus haut)
n_samples, n_features = X_poly.shape
W = np.zeros(n_features)
b = 0
lr = 0.001 # Attention : x^2 grandit vite, donc petit lr

for i in range(5000):
    y_pred = np.dot(X_poly, W) + b
    dw = (2/n_samples) * np.dot(X_poly.T, (y_pred - y))
    db = (2/n_samples) * np.sum(y_pred - y)
    W = W - lr * dw
    b = b - lr * db

print(f"Coefficients x et x^2: {W}")
```
