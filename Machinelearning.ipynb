```python
import numpy as np

# Données d'entraînement
# X = variables d'entrée (features)
# y = valeurs réelles (target)
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

# Initialisation des paramètres du modèle
# a = coefficient (pente de la droite)
# b = biais (intercept)
a, b = 0.0, 0.0

# hyperparamètres
lr = 0.1              # learning rate (taille du pas d'apprentissage)
epochs = 2000         # nombre maximum d'itérations
epsilon = 1e-6        # seuil pour arrêter si amélioration est très petite

# coût précédent (initialisé à infini)
previous_cost = float('inf')

# boucle d'entraînement (Gradient Descent)
for i in range(epochs):

    #prédiction du modèle actuel
    y_pred = a * X + b

    #calcul de la fonction de coût (MSE)
    current_cost = (1/len(X)) * sum((y - y_pred)**2)

    # condition d'arrêt (convergence)
    # si l'amélioration du coût est très petite → stop
    if abs(previous_cost - current_cost) < epsilon:
        break

    # 🔄 mise à jour du coût précédent
    previous_cost = current_cost

    # 📊 calcul des gradients (dérivées)
    # da = direction d'ajustement de a
    # db = direction d'ajustement de b
    da = (-2/len(X)) * sum(X * (y - y_pred))
    db = (-2/len(X)) * sum(y - y_pred)

    # 🔄 mise à jour des paramètres (gradient descent)
    a = a - lr * da
    b = b - lr * db

#  affichage du modèle final appris
print(f"y = {a:.2f}x + {b:.2f}")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# 1. Préparation des données
# X doit toujours être une matrice 2D (d'où le .reshape(-1, 1))
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 5])

# 2. Initialisation du modèle
model = LinearRegression()

# 3. Entraînement (Le "fit")
# C'est ici que l'algorithme calcule a et b automatiquement
model.fit(X, y)

# 4. Prédiction
# On demande au modèle de prédire pour une nouvelle valeur (ex: 6)
nouvelle_valeur = np.array([[6]])
prediction = model.predict(nouvelle_valeur)

print(f"Pente (a) : {model.coef_[0]:.2f}")
print(f"Ordonnée à l'origine (b) : {model.intercept_:.2f}")
print(f"Prédiction pour x=6 : {prediction[0]:.2f}")

# (Optionnel) Visualisation
plt.scatter(X, y, color='blue') # Points réels
plt.plot(X, model.predict(X), color='red') # Ligne de régression
plt.title("Régression Linéaire avec Scikit-Learn")
plt.show()
```

```python
import numpy as np

#Données d'entraînement
# X = features (variables d'entrée)
# Ici: [Surface, Nombre de chambres]
X = np.array([
    [50, 2],
    [80, 3],
    [100, 4],
    [30, 1],
    [120, 5]
])

# y = target (prix de la maison)
y = np.array([200, 310, 390, 130, 480])

# n_samples = nombre d'exemples
# n_features = nombre de variables (features)
n_samples, n_features = X.shape

# W = poids (coefficients) pour chaque feature
# ici: W = [w1, w2]
W = np.zeros(n_features)

# b = biais (intercept)
b = 0

# lr = learning rate (taille des pas d'apprentissage)
lr = 0.0001

# epochs = nombre d'itérations
epochs = 1000

# TRAINING LOOP
for i in range(epochs):

    # PRÉDICTION DU MODÈLE
    # produit scalaire: X·W + b
    y_pred = np.dot(X, W) + b

    # FONCTION DE COÛT (MSE)
    # mesure l'erreur du modèle
    cost = (1/n_samples) * np.sum((y_pred - y)**2)

    # CALCUL DES GRADIENTS
    # dw = dérivées par rapport aux poids W
    dw = (2/n_samples) * np.dot(X.T, (y_pred - y))

    # db = dérivée par rapport au biais b
    db = (2/n_samples) * np.sum(y_pred - y)

    # MISE À JOUR DES PARAMÈTRES
    # on ajuste W et b pour réduire l'erreur
    W = W - lr * dw
    b = b - lr * db


# RÉSULTAT FINAL

print(f"Poids (W): {W}, Biais (b): {b:.2f}")
```

```python
import numpy as np

# Données d'entraînement
# X = variable d'entrée (non linéaire)
# y = cible (ici relation courbe type x²)
X = np.array([1, 2, 3, 4, 5])
y = np.array([1.2, 4.5, 8.8, 16.2, 24.8])

# TRANSFORMATION POLYNOMIALE

# On transforme X en features polynomiales:
# x → [x, x²]
X_poly = np.column_stack((X, X**2))

#INITIALISATION


# n_samples = nombre d'exemples
# n_features = nombre de colonnes après transformation
n_samples, n_features = X_poly.shape

# W = poids (coefficients pour x et x²)
W = np.zeros(n_features)

# b = biais (intercept)
b = 0

# lr = learning rate
# petit car x² augmente vite
lr = 0.001


#TRAINING (Gradient Descent)


for i in range(5000):

    #  PRÉDICTION DU MODÈLE
    # y = W1*x + W2*x² + b
    y_pred = np.dot(X_poly, W) + b

    #  FONCTION DE COÛT (MSE dérivée)
    # mesure l'erreur entre prédiction et vraie valeur
    cost = (1/n_samples) * np.sum((y_pred - y)**2)

    #  CALCUL DES GRADIENTS
    # dérivées pour ajuster W et b
    dw = (2/n_samples) * np.dot(X_poly.T, (y_pred - y))
    db = (2/n_samples) * np.sum(y_pred - y)

    #  MISE À JOUR DES PARAMÈTRES
    W = W - lr * dw
    b = b - lr * db


#  RÉSULTAT FINAL

print(f"Coefficients x et x^2: {W}")
```

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

#  Données (non linéaires)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([1.2, 4.5, 8.8, 16.2, 24.8])


# 1 Transformation polynomiale

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

# X_poly = [1, x, x²]
# column 0 = bias (1)
# column 1 = x
# column 2 = x²


# 2 Model Linear Regression

model = LinearRegression()

# training
model.fit(X_poly, y)


# 3 Predictions

y_pred = model.predict(X_poly)


# 4 Results

print("Intercept (b):", model.intercept_)
print("Coefficients (w1, w2):", model.coef_)
```
